```python
import pandas as pd
pd.set_option('display.max_rows', None)
df_elec = pd.read_csv('FPL.txt', low_memory = False)
df_weather = pd.read_csv('MIA_2025.txt', low_memory = False)
```

```python
# Gather all the elevations into a new list
allElevations = []
HourlySkyConditions = list(df_weather['HourlySkyConditions'])

# First create a new column with the elevation
for condition in HourlySkyConditions:
    condition = str(condition)
    if len(condition) < 4:
        allElevations.append(None)
        continue
    # Elevation always starts on the 7th charecter
    elevation = ''
    start = 7
    end = condition.find(' ', start + 1)
    if end == -1:
        elevation = condition[start::]
    else:
        elevation = condition[start:end]
    if elevation == '':
        elevation = '120'
    if elevation[-1] == 's':
        elevation = elevation.rstrip().lstrip()
        elevation = elevation[:-1]
    allElevations.append(int(elevation))
```

```python
allCloudCoverage = []
HourlySkyConditions = list(df_weather['HourlySkyConditions'])

for condition in HourlySkyConditions:
    condition = str(condition)
    coverage = condition[4:6]
    if coverage == '':
        allCloudCoverage.append(None)
        continue
    allCloudCoverage.append(int(coverage))
```

```python
df_weather['Elevation'] = allElevations
df_weather['Oktas'] = allCloudCoverage
one_hot_encoding = pd.get_dummies(df_weather.Oktas)
one_hot_encoding.drop(9.0, axis=1)
```
